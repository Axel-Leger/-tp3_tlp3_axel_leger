### Actividad 5:
Sumar los elementos de dos listas posición por posición, utilizando solo bucles repetititivos (for,while).

```python

# todo:

lists_one = [ 2,4,6,7,8 ]
lists_two = [ 1,3,5,9,10 ]
resultado = []

for numero in range(len(lists_one)):
    suma = lists_one[numero] + lists_two[numero]
    resultado.append(suma)

print("El resultado de la suma es: ", resultado)
```

### Actividad 6:

Escribir un programa que almacene el abecedario en una lista, elimine de la lista las letras que ocupen posiciones múltiplos de 3, y muestre por pantalla la lista resultante.

```python

# todo:

abecedario = list("abcdefghijklmnopqrstuvwxyz")
resultado = []

for letra in range(len(abecedario)):
    if (letra + 1) % 3 != 0:
        resultado.append(abecedario[letra])

print("Abecedario sin letras múltiplos de 3:", resultado)
```
